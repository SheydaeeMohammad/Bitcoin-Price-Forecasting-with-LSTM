```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM
from tensorflow.keras.layers import Dense, Activation, Input, Dropout
from keras.optimizers import Adam
```

```python
import pandas as pd
```

```python
train = pd.read_csv("data\Train_BTC.csv")
test = pd.read_csv("data\Test_BTC.csv")
```

```python
print(train.iso_month.nunique())
print(test.iso_month.nunique())
```

```python
period_length = 4
number_of_periods = 108 - 1
print(number_of_periods)
```

```python
def build_model(period_length, number_of_periods, batch_size=1):
  
    model = Sequential()
    model.add(Input(shape=(number_of_periods, period_length)))
    model.add(LSTM(units=period_length, activation="relu", dropout=0.2, return_sequences=False, stateful=False))
    model.add(Dense(units=period_length, activation="relu"))

    model.compile(optimizer=Adam(learning_rate=0.001), loss="mean_squared_error")

    return model
```

```python
model = build_model(period_length=period_length, number_of_periods=number_of_periods)
```

```python
model.save('real_bitcoin_lstm_v0_me.keras')
```
